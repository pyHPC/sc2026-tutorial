### Introduction

This tutorial starts with NumPy and the `ndarray` memory model, moves the same workflows to the GPU with CuPy, covers memory spaces and data movement between CPU and GPU, profiles asynchronous execution, and distributes work across processes with mpi4py.

We explore an HPC mini-application solving the Euler characteristic curve of a cosmic-web density field: 3D stencil followed by a histogram, implemented with CuPy, JAX, cuda.compute, Numba CUDA, and CppJIT. We then go over examples of scientific applications that interoperate efficiently between Python and C++. The day closes with a session on cuPhoton and DragonHPC (placeholder).

In this tutorial we will cover:
- NumPy fundamentals and the `ndarray` memory model.
- Accelerating array workflows on the GPU with CuPy, including memory spaces and asynchrony.
- Distributed computing with mpi4py for multi-node parallelism.
- Programming models in Python for GPU applications: array expressions, fusion, library iterators, kernel authoring, and interoperability with CUDA C++ kernels.
- Reading an application's profile with Nsight Systems and Nsight Compute.
- Case studies of horizontal Python and C++ integration in domain sciences such as cancer research and high energy physics.
- DragonHPC fundamentals enabling `multiprocessing` to scale across HPC multi-node resources.
- cuPhoton astronomical image processing workflows orchestrated using DragonHPC.

Attendees are expected to have general knowledge of HPC and Python programming. Some familiarity with NumPy and C/C++ is helpful but not required.
